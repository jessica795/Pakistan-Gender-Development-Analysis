# Pakistan Gender & Development Analysis

## 02 — Data Cleaning

This notebook prepares the World Bank dataset for statistical analysis.

The analysis focuses on Pakistan during 2006–2024, using indicators with sufficient overlapping observations.

### Main outcome

Female labor force participation rate (% of female population ages 15+)

### Explanatory variables

- Female secondary school enrollment
- GDP per capita
- Fertility rate
- Urban population
- Life expectancy

Missing observations are not artificially imputed. Years without complete observations are excluded from the analysis dataset.

```python
import pandas as pd
```

```python
import pandas as pd

df = pd.read_csv("/content/pakistan_gender_development_raw.csv")

df.head()
```

```python
selected_variables = [
    "Female_Labor_Force_Participation",
    "Female_Secondary_Enrollment",
    "GDP_per_Capita",
    "Fertility_Rate",
    "Urban_Population",
    "Life_Expectancy"
]

df_clean = df[
    ["Country", "Country_Code", "Year"] + selected_variables
].copy()
```

```python
df_clean = df_clean[
    (df_clean["Year"] >= 2006) &
    (df_clean["Year"] <= 2024)
].copy()

df_clean = df_clean.sort_values("Year").reset_index(drop=True)

df_clean
```

```python
selected_variables = [
    "Female_Labor_Force_Participation",
    "Female_Secondary_Enrollment",
    "GDP_per_Capita",
    "Fertility_Rate",
    "Urban_Population",
    "Life_Expectancy"
]

df_complete = df_clean.dropna(
    subset=selected_variables
).copy()

print("Dataset shape:", df_complete.shape)
print("\nDuplicate rows:", df_complete.duplicated().sum())
print("Duplicate years:", df_complete["Year"].duplicated().sum())
```

```python
df_complete[selected_variables].describe().T
```

```python
print("Dataset shape:", df_complete.shape)

print("\nDuplicate rows:", df_complete.duplicated().sum())
print("Duplicate years:", df_complete["Year"].duplicated().sum())

print("\nData types:")
print(df_complete.dtypes)
```

```python
df_complete.to_csv(
    "pakistan_gender_development_processed.csv",
    index=False
)

print("Processed dataset saved successfully.")
print("Final dataset shape:", df_complete.shape)
```
